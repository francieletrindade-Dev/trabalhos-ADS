```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Criando um DataFrame com o número de alunos por titulação
dados = {
    'Titulação': ['Doutorado', 'Mestrado', 'Especialização', 'Graduação', 'Cursos Técnicos'],
    'Número de Alunos': [20, 15, 85, 145, 320]
}

df = pd.DataFrame(dados)

# Exibir a tabela
print("Tabela de Titulações e Número de Alunos:\n")
print(df)
print("\nResumo Estatístico:\n")
print(df['Número de Alunos'].describe())

# Calcular porcentagem
df['% do Total'] = (df['Número de Alunos'] / df['Número de Alunos'].sum()) * 100

# Criar gráfico de barras
plt.figure(figsize=(8,5))
bars = plt.bar(df['Titulação'], df['Número de Alunos'], color=['#4F81BD', '#C0504D', '#9BBB59', '#8064A2', '#F79646'], edgecolor='black')

# Adicionar rótulos de valores e porcentagens
for i, bar in enumerate(bars):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 5, f"{df['% do Total'][i]:.1f}%", ha='center', fontsize=9)

plt.title('Número de Alunos por Titulação', fontsize=14, fontweight='bold')
plt.xlabel('Titulação')
plt.ylabel('Número de Alunos')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Criando o DataFrame
df = pd.DataFrame({
    'Titulação': ['Doutorado', 'Mestrado', 'Especialização', 'Graduação', 'Cursos Técnicos'],
    'Quantidade': [20, 15, 85, 145, 320]
})

# Exibir a tabela formatada
print("Tabela - Número de Alunos por Titulação:\n")
print(df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# Definindo variáveis
fatias = [8, 12, 10]                       # Valores de cada categoria
atividades = ['Lazer', 'Estudo', 'Viagem'] # Legendas
cores = ['r', 'm', 'y']                    # Cores: vermelho, magenta e amarelo

# Criando o gráfico de pizza
plt.pie(
    fatias,
    labels=atividades,
    colors=cores,
    startangle=90,     # Começa o gráfico a partir de 90°
    shadow=True,       # Adiciona sombra
    explode=(0, 0.1, 0),  # Destaca o segundo item (Estudo)
    autopct='%1.1f%%'  # Mostra os percentuais
)

plt.title('Distribuição de Atividades')  # Título do gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt

# Definindo variáveis
x = [1, 3, 5]
y = [1, 2, 5]

# Criando o gráfico de linha com legenda
plt.plot(x, y, label='Uma legenda', color='b', marker='o', linestyle='-', linewidth=2)

# Adicionando título e rótulos aos eixos
plt.title('Exemplo utilizando Plot')
plt.xlabel('Variável 1')
plt.ylabel('Variável 2')

# Exibindo a legenda
plt.legend()

# Exibindo o gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt

# Definindo variáveis
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [5, 2, 4, 5, 6, 8, 4, 8]

# Criando o gráfico de dispersão
plt.scatter(
    x, y,
    label='Pontos',    # legenda
    color='b',         # cor azul
    marker='*',        # formato de estrela
    s=100              # tamanho dos pontos
)

# Adicionando título e rótulos
plt.title('Gráfico de Pontos em Formato de Estrela')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

# Exibindo legenda e grade
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Exibindo gráfico
plt.show()
```
